# Kaggle Color Tester

This notebook is to test the accuracy of the colors we added to the kaggle dataset. 

```python
import json
import random
from pathlib import Path
import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent.parent
DATA_PATH    = PROJECT_ROOT / "fashion_items" / "styles.json"
IMAGE_DIR    = PROJECT_ROOT / "fashion_items" / "images"
```

```python
with open(DATA_PATH) as f:
    items = json.load(f)

# ── Display one item ──────────────────────────────────────────────────────────
def show_item(item):
    colors = item.get('dominant_color', []), item.get('secondary_color', [])
    img_path = IMAGE_DIR / f"{item['id']}.jpg"


    if img_path.exists():
        img_widget = widgets.Image(value=open(img_path, "rb").read(), width=300)
    else:
        img_widget = widgets.Label("No image found")


    swatches = "".join(
        f'<div style="display:flex; align-items:center; margin:4px 0">'
        f'  <div style="width:40px; height:40px; border-radius:6px; background:rgb({c[0]},{c[1]},{c[2]}); margin-right:10px"></div>'
        f'  <code>rgb({c[0]},{c[1]},{c[2]})</code>'
        f'</div>'
        for c in colors
    )
    colors_widget = widgets.HTML(f"<b>ID: {item['id']}, Type: {item['articleType']}</b><br><br>{swatches}")

    with out:
        clear_output(wait=True)
        display(widgets.HBox([img_widget, colors_widget]))


```

```python
out = widgets.Output()
btn = widgets.Button(description="Next Random Item", button_style="primary",
                     layout=widgets.Layout(width="200px", height="40px"))

btn.on_click(lambda _: show_item(random.choice(items)))
display(btn, out)
show_item(random.choice(items))
```
