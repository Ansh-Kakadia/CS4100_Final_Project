# Kaggle Color Tester

This notebook is to test the accuracy of the colors we added to the kaggle dataset. 

```python
import json
import random
import ast
from pathlib import Path

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

DATA_PATH   = "fashion_items/styles.json"
IMAGE_DIR   = "fashion_items/images/"
```

```python
with open(DATA_PATH) as f:
    items = json.load(f)

# ── Display one item ──────────────────────────────────────────────────────────
def show_item(item):
    colors = item.get('dominant_color', []), item.get('secondary_color', [])
    img_path = Path(IMAGE_DIR) / item.get('id', "")

    # Image widget
    if img_path.exists():
        img_widget = widgets.Image(value=open(img_path, "rb").read(), width=300)
    else:
        img_widget = widgets.Label("No image found")

    # Color swatches as HTML
    swatches = "".join(
        f'<div style="display:flex; align-items:center; margin:4px 0">'
        f'  <div style="width:40px; height:40px; border-radius:6px; background:{c}; margin-right:10px"></div>'
        f'  <code>{c.upper()}</code>'
        f'</div>'
        for c in colors
    )
    colors_widget = widgets.HTML(f"<b>ID: {item['id']}</b><br><br>{swatches}")

    with out:
        clear_output(wait=True)
        display(widgets.HBox([img_widget, colors_widget]))

# ── Widget ────────────────────────────────────────────────────────────────────
out = widgets.Output()
btn = widgets.Button(description="🔀 Next Random Item", button_style="primary",
                     layout=widgets.Layout(width="200px", height="40px"))

btn.on_click(lambda _: show_item(random.choice(items)))
display(btn, out)
show_item(random.choice(items))
```
